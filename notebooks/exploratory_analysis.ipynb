```python
import os
import matplotlib.pyplot as plt
from PIL import Image

# 1. Define paths
DATA_DIR = "../PlantVillage/"  # Adjust based on where your notebook sits
classes = os.listdir(DATA_DIR)

# 2. Check Class Distribution (Bar Chart)
counts = {cls: len(os.listdir(os.path.join(DATA_DIR, cls))) for cls in classes}
plt.bar(counts.keys(), counts.values(), color='teal')
plt.title("Class Distribution - PlantVillage Subset")
plt.xlabel("Crop Condition")
plt.ylabel("Number of Images")
plt.xticks(rotation=15)
plt.savefig("../outputs/class_distribution.png")  # Save for your slides!
plt.show()

# 3. Plot Sample Images
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for i, cls in enumerate(classes[:3]):
    img_name = os.listdir(os.path.join(DATA_DIR, cls))[0]
    img_path = os.path.join(DATA_DIR, cls, img_name)
    img = Image.open(img_path)
    axes[i].imshow(img)
    axes[i].set_title(cls)
    axes[i].axis('off')
plt.show()
```
