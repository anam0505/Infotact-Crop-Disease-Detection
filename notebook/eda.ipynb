```python
import os
import matplotlib.pyplot as plt
from PIL import Image
```

```python
train_path = "../data/raw/Apple/Train"
val_path = "../data/raw/Apple/Val"
test_path = "../data/raw/Apple/Test"
```

```python
classes = os.listdir(train_path)

for cls in classes:
    img_path = os.path.join(train_path, cls, os.listdir(os.path.join(train_path, cls))[0])
    img = Image.open(img_path)

    plt.imshow(img)
    plt.title(cls)
    plt.axis('off')
    plt.show()
```

```python
counts = {}

for cls in os.listdir(train_path):
    counts[cls] = len(os.listdir(os.path.join(train_path, cls)))

plt.bar(counts.keys(), counts.values())
plt.xticks(rotation=45)
plt.title("Training Data Class Distribution")
plt.xlabel("Classes")
plt.ylabel("Number of Images")
plt.show()
```

```python
val_counts = {}

for cls in os.listdir(val_path):
    val_counts[cls] = len(os.listdir(os.path.join(val_path, cls)))

print("Validation Data:")
print(val_counts)
```

```python
sizes = []

for cls in os.listdir(train_path):
    img_path = os.path.join(train_path, cls, os.listdir(os.path.join(train_path, cls))[0])
    img = Image.open(img_path)
    sizes.append(img.size)

print("Sample Image Sizes:", sizes)
```

```python
import random

for cls in classes:
    images = os.listdir(os.path.join(train_path, cls))
    sample = random.sample(images, min(3, len(images)))

    for img_name in sample:
        img_path = os.path.join(train_path, cls, img_name)
        img = Image.open(img_path)

        plt.imshow(img)
        plt.title(cls)
        plt.axis('off')
        plt.show()
```
